```python
# ==========================================================
# 🧹 ALIGNERR - DATA CLEANER & VALIDATOR TOOL
# Dibuat oleh: Raisya Putri Nasywa
# Deskripsi: Program untuk Membersihkan & Memeriksa Kualitas Data AI
# ==========================================================

def bersihkan_teks(teks_input):
    """Fungsi membersihkan teks/data yang kotor atau berantakan"""
    teks_bersih = teks_input.strip()       # Hapus spasi pinggir
    teks_bersih = teks_bersih.lower()      # Jadi huruf kecil semua
    teks_bersih = " ".join(teks_bersih.split()) # Hapus spasi ganda

    # Rapikan tanda baca
    teks_bersih = teks_bersih.replace(" ,", ",")
    teks_bersih = teks_bersih.replace(" .", ".")
    teks_bersih = teks_bersih.replace(" !", "!")
    teks_bersih = teks_bersih.replace(" ?", "?")
    return teks_bersih


def cek_kualitas(teks):
    """Fungsi memeriksa apakah data bagus atau ada masalah"""
    if len(teks) == 0:
        return "❌ GAGAL: Data kosong / tidak ada isinya!"
    elif len(teks) < 3:
        return "⚠️ PERINGATAN: Data terlalu pendek!"
    else:
        return "✅ BAGUS: Data bersih & siap diproses AI!"


# ==========================================================
# 🚀 BAGIAN UTAMA - JALANKAN PROGRAM
# ==========================================================
if __name__ == "__main__":
    print("=" * 50)
    print("🧹  PROGRAM PEMBERSIH & PENGECEK DATA AI  🧹")
    print("   Khusus untuk Alignerr")
    print("=" * 50)

    # Contoh data kotor yang mau dibersihkan
    daftar_data = [
        "   Saya  sedang  bekerja  di  Alignerr   ",
        "",  # Data kosong
        "   Jawaban  ini  sudah  BENAR  !!   ",
        "   AI adalah  teknologi  masa depan   ",
        "   a   "
    ]

    nomor = 1
    for data in daftar_data:
        print(f"\n📝 DATA KE-{nomor}:")
        print(f"Awal  : '{data}'")
        hasil = bersihkan_teks(data)
        print(f"Bersih: '{hasil}'")
        print(f"Status: {cek_kualitas(hasil)}")
        nomor += 1

    print("\n" + "=" * 50)
    print("✅ SELESAI! SEMUA DATA SUDAH SIAP PAKAI!")
    print("=" * 50)
```
